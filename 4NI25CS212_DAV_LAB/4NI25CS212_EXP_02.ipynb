```python
print("Shape of dataset (rows,columns):",df.shape)
print("\nColumn names:",df.columns.tolist())
print("\nData types of each column:",df.dtypes)
```

```python
print("\n\nInformation : \n")
df.info()
print("\n\nDiscriptive Statistics (numerical columns):")
print(df.describe())
print("\nMean : \n",df.mean(numeric_only=True))
print("\nMedian :\n",df.median(numeric_only=True))
print("\nMode :\n",df.mode(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\n\nNull Entries : \n")
print(df.isnull())
print("\nTotal Number of Null Entries : \n")
print(df.isnull().sum())
print("\nTotal missing values in dataset :",df.isnull().sum().sum())
```
